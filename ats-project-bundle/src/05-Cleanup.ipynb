```python
class Cleanup:
    def __init__(self):
        pass

    def cleanup_metadata(self):
        print("Deleting the metadata ....", end ="")
        spark.sql(f"drop table if exists {conf.jobs_metadata_table_name}")
        print("done")
    
    def cleanup_landingzone(self):
        print("Deleting the landingzone directories ....", end ="")
        profile_landing_zone = (
            f"/Volumes/{conf.catalog}/{conf.db}/{conf.profile_landing_zone}"
        )
        dbutils.fs.rm(profile_landing_zone,recurse=True)
        jd_landing_zone = f"/Volumes/{conf.catalog}/{conf.db}/{conf.jd_landing_zone}"
        dbutils.fs.rm(jd_landing_zone, recurse=True)
        print("done")

    def cleanup_tables(self):
        print (f"Deleting Tables...", end="")
        spark.sql(f"drop table if exists {conf.profile_bronze_table_name}")
        spark.sql(f"drop table if exists {conf.profile_silver_table_name}_stg")
        spark.sql(f"drop table if exists {conf.profile_silver_table_name}")
        spark.sql(f"drop table if exists {conf.jd_bronze_table_name}")
        spark.sql(f"drop table if exists {conf.jd_silver_table_name}_stg")
        spark.sql(f"drop table if exists {conf.jd_silver_table_name}")
        spark.sql(f"drop table if exists {conf.jd_profile_table_name}_stg")
        spark.sql(f"drop table if exists {conf.jd_profile_table_name}")
        print("Done")

    def cleanup_endpoints(self):
        ep = Endpoints()
        ep.delete(conf.llm_endpoint_for_chat)

    def cleanup_vector_index(self):
        import time
        vp = VectorSearch()
        vp.delete_index(conf.vector_index_name,conf.vector_store_endpoint_name)
        time.sleep(10)
        vp.delete_endpoint(conf.vector_store_endpoint_name)

    
    def run(self):
        self.cleanup_metadata()
        self.cleanup_landingzone()
        self.cleanup_tables()
        self.cleanup_vector_index()
        self.cleanup_endpoints()


```

```python
cleanup = Cleanup()
cleanup.run()
```
